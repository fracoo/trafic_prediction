<h2>Description</h2>

Dans ce code, nous traitons le dataset relatif aux vacances parisiennes. 
Les données proviennent de https://data.education.gouv.fr/explore/dataset/fr-en-calendrier-scolaire/table/?disjunctive.zones&disjunctive.description&disjunctive.location&disjunctive.annee_scolaire&disjunctive.population&sort=end_date

Nous voulons obtenir un dataframe avec les dates de début et de fin des vacances.

Ces données peuvent a priori avoir une incidence directe sur la circulation parisienne, donc nous allons les inclure dans nos bases de données. 

```python
import pandas as pd

print("Version installée de pandas :")
print(pd.__version__)
```

```python
df_vacances = pd.read_csv('../datasets_externes_bruts/fr-en-calendrier-scolaire.csv', sep=';')

df_vacances['Date de début'] = df_vacances['Date de début'].apply(lambda x: x[:10])
df_vacances['Date de fin'] = df_vacances['Date de fin'].apply(lambda x: x[:10])

df_vacances['Date de début'] = pd.to_datetime(df_vacances['Date de début'])
df_vacances['Date de fin'] = pd.to_datetime(df_vacances['Date de fin'])

df_vacances.info()
```

```python
df_vacances = df_vacances[df_vacances['Population'] != 'Enseignants']
df_vacances = df_vacances.drop('Population', axis=1)

df_vacances = df_vacances[df_vacances['Académies'] == "Paris"]

df_vacances = df_vacances[(df_vacances['Date de début'] > pd.to_datetime("01/01/2024")) 
                           & (df_vacances['Date de fin'] < pd.to_datetime("12/31/2025"))]

df_vacances = df_vacances.sort_values(by="Date de début", ascending=True).reset_index(drop=True)

df_vacances.head(20)
```

```python
df_vacances.to_csv("../datasets_externes_clean/vacances.csv", sep=";")
```
