<h2>Description</h2>

Dans ce code, nous traitons le dataset relatif aux jours fériés.
Les données qui nous intéressent dans cette base sont la date des jours feriés. 

```python
import pandas as pd

print("Version installée de pandas :")
print(pd.__version__)
```

```python
df_ferie = pd.read_csv('../datasets_externes_bruts/ferie.csv', sep=";")

df_ferie.info()
```

```python
df_ferie['datetime'] = pd.to_datetime(df_ferie['Date de début'], format='%d/%m/%Y',)

df_ferie = df_ferie.sort_values(by="datetime", ascending=True)

df_ferie = df_ferie[(df_ferie['datetime'] > pd.to_datetime('01/01/2024')) & (df_ferie['datetime'] < pd.to_datetime('12/31/2025'))].reset_index(drop=True)

df_ferie.head(20)
```

```python
df_ferie.to_csv("../datasets_externes_clean/ferie.csv", sep=";")
```
