<h2>Description</h2>

L'objectif de ce code est d'inclure dans nos bases de données les informations relatives aux dimanches constituant l'opération "Paris respire". 

Cet évènement influence fortement le trafic en ces dimanches. 

```python
import pandas as pd 

print("Version installée de pandas :")
print(pd.__version__)
```

```python
# Liste des dimanches piétons (premier dimanche du mois généralement)
dimanches_pietons = [
    # 2024
    '2024-09-01',
    '2024-10-06',
    '2024-11-03',
    '2024-11-11',
    '2024-12-01',
    # 2025
    '2025-01-05',
    '2025-02-02',
    '2025-03-02',
    '2025-04-06',
    '2025-05-04',
    '2025-06-01',
    '2025-07-06',
    '2025-08-03',
    '2025-09-07',
    '2025-10-05',
    '2025-11-11',
    '2025-09-21',  # Exceptionnellement
    '2025-09-22',  # Exceptionnellement
    '2025-11-16',  # Exceptionnellement
    '2025-12-07'   # Exceptionnellement
]

dimanches_pietons = pd.to_datetime(dimanches_pietons)

def get_horaires_pietonnisation(date):
    """
    Retourne les horaires de piétonnisation selon la saison:
    - Avril à octobre: 11h à 18h
    - Novembre à mars: 10h à 17h
    """
    mois = date.month
    jour = date.day
    annee = date.year
    if mois == 9 and jour == 21 and annee == 2025:
        return 5,23
    if mois == 9 and jour == 22 and annee == 2025:
        return 0, 1
    if mois == 11 and jour == 11:
        return 7, 14
    if 4 <= mois <= 10:
        return 11, 18
    else: 
        return 10, 17

df_pieton = pd.DataFrame({'date': dimanches_pietons})

df_pieton['date_debut'] = df_pieton['date'].apply(lambda d: pd.Timestamp(d.date()) + pd.Timedelta(hours=get_horaires_pietonnisation(d)[0]))
df_pieton['date_fin'] = df_pieton['date'].apply(lambda d: pd.Timestamp(d.date()) + pd.Timedelta(hours=get_horaires_pietonnisation(d)[1]))

df_pieton = df_pieton.drop('date', axis=1)

df_pieton = df_pieton.sort_values(by="date_debut", ascending=True).reset_index(drop=True)

df_pieton.head(20)
```

```python
df_pieton.to_csv("../datasets_externes_clean/pieton.csv", sep=";")
```
