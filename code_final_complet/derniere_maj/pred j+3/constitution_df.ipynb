```python
import pandas as pd
import numpy as np
```

```python
dates = pd.date_range(
    start="2025-11-09 00:00:00",
    end="2025-11-12 00:00:00",
    freq="H",
    inclusive="left"   
)

df_axe= pd.DataFrame({'datetime': dates})

df_axe['date'] = df_axe['datetime'].dt.date
df_axe['heure'] = df_axe['datetime'].dt.hour
df_axe['dow'] = df_axe['datetime'].dt.dayofweek
df_axe['mois'] = df_axe['datetime'].dt.month
df_axe['annee'] = df_axe['datetime'].dt.year
df_axe['jour_mois'] = df_axe['datetime'].dt.day
jour_map = {0:'Lun',1:'Mar',2:'Mer',3:'Jeu',4:'Ven',5:'Sam',6:'Dim'}
df_axe['jour_semaine'] = df_axe['dow'].map(jour_map)

df_axe['heure_sin'] = np.sin(2 * np.pi * df_axe['heure'] / 24)
df_axe['heure_cos'] = np.cos(2 * np.pi * df_axe['heure'] / 24)

df_axe['jour_sin'] = np.sin(2 * np.pi * df_axe['dow'] / 7)
df_axe['jour_cos'] = np.cos(2 * np.pi * df_axe['dow'] / 7)

df_axe['mois_sin'] = np.sin(2 * np.pi * df_axe['mois'] / 12)
df_axe['mois_cos'] = np.cos(2 * np.pi * df_axe['mois'] / 12)

df_axe.head(10)
```

```python
prev_meteo = pd.read_csv("previsions_meteo.csv" , sep=";")

prev_meteo['time'] = pd.to_datetime(prev_meteo['time'])

df_merge = pd.merge(df_axe, prev_meteo, left_on='datetime', right_on='time', how='left')

df_merge = df_merge.rename(columns={'temperature_2m (¡C)' : 'Température',
                                    'precipitation (mm)': 'precipitations heure',
                                    'Vent (m/s)' : 'force moyenne vent (m/s)'})
df_merge = df_merge.drop('Unnamed: 0',axis = 1)

df_merge.head()
```

```python
df = df_merge

def est_weekend(jour):
    if jour == "Dim" or jour == "Sam":
        return 1
    return 0

df['est_weekend'] = df['jour_semaine'].apply(est_weekend)

def est_ferie(num):
    if num == 11:
        return 1
    return 0

df['est_ferie'] = df['jour_mois'].apply(est_ferie)

def est_av_ferie(num):
    if num == 10:
        return 1
    return 0

df['est_avant_ferie'] = df['jour_mois'].apply(est_av_ferie)

df['est_vacances'] = np.nan
df['est_vacances'] = df['est_vacances'].fillna(0)

df['est_avant_vacances'] = np.nan
df['est_avant_vacances'] = df['est_vacances'].fillna(0)

df['est_rentree'] = np.nan
df['est_rentree'] = df['est_rentree'].fillna(0)

def est_pieton(date) :
    date = pd.to_datetime(date)
    inf = pd.to_datetime('2025-11-11 06:00:00')
    sup = pd.to_datetime('2025-11-11 14:00:00')
    if date >= inf and date <= sup :
        return 1
    return 0

df['est_pieton'] = df['datetime'].apply(est_pieton)

def est_hpointe(date):
    heure = date.hour
    if heure == 18 or heure == 19:
        return 1
    return 0

df['est_hpointe'] = df['datetime'].apply(est_hpointe)

df.head(50)

```

```python
df['Température'] = df['Température'].str.replace(',','.')
df['precipitations heure'] = df['precipitations heure'].str.replace(',' , '.')
df['force moyenne vent (m/s)'] = df['force moyenne vent (m/s)'].str.replace(',','.')
df['Température'] = pd.to_numeric(df['Température']).astype(float)
df['precipitations heure'] = pd.to_numeric(df['precipitations heure']).astype(float)
df['force moyenne vent (m/s)'] = pd.to_numeric(df['force moyenne vent (m/s)']).astype(float)
```
